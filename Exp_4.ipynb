```python
#Set is unordered.
set1={"apple", "banana", "cherry"}
print(set1)
```

```python
#Indexing commands don't work in set.
set1={"apple", "banana", "cherry"}
print(set1[1])
```

```python
#Data types can be put in a set but is still unordered.
set2={2,2.5,True,"Dhruv"}
print(set2)
```

```python
#set removes duplicates.
set3={1,2.4,False,"hello",1}
print(set3)
```

```python
#Addition of elements is allowed unlike tupple.
#set is mutable.
set4={"Apple", "Orange", "Banana"}
set4.add("Mango")
print(set4)
```

```python
#Removing of elements is allowed.
set5={"Apple", "Orange", "Banana", "Mango"}
set5.remove("Mango")
print(set5)
```

```python
set6={"Apple", "Orange", "Banana", "Mango"}
set7={"Apple", "Orange", "Banana","Kiwi"}
print("Union:", set6|set7)       #taking non-different elements of 2 sets.
print("Intersection:", set6&set7)#taking common elements of both sets.
print("Difference_1:", set6-set7)#taking elements of set6 which are not in set7.
print("Difference_2:", set7-set6)#taking elements of set7 which are not in set6.
print("*XOR Difference3:", set6^set7)#adding + removing the common elements.
```

```python
set8=frozenset({"Apple", "Orange", "Banana", "Mango"})
print(set8)
print(type(set8))
```

```python
'''problem1: unique Event participants
A college event has a list of students who registered multiple times by mistake
Write a program using set to store participant names and '''
x=["Rishabh", "Rishabh", "Ronit", "Soham", "Dhruv"]
print("participants:", set(x))
```

```python
'''Problem2: Three students choose elective subjects.
Find the common subjects chosen by all students using set operation.
Concept used: Set Intersection (&).'''

student1={"python", "eg", "math", "dcld"}
student2={"python", "hindi", "english", "Sanskrit"}
student3={"python", "tpcs", "pes", "java"}
common_subjects=student1&student2&student3
print("Common Subjects:", common_subjects)
```

```python
'''Problem3: One set contains students in the cricket club
another in the football club.
Find students who are: 1. In both clubs.
2. In only 1 club.
concept used: Intersection, Symmetric difference.'''
cricket={"Dhruv", "Raghav", "Rishabh"}
football={"Dhruv", "Rishabh", "Ronit"}
print("Both clubs:", cricket & football)
print("In 1 club :", cricket ^ football)
```

```python
'''problem4: Absent students
Given a set of all students and a set of present students,
find the absent students.'''
students={"Dhruv", "Raghav", "Rishabh", "Ronit"}
present={"Dhruv", "Rishabh"}
absent=students-present
print("Absent students:", absent)
```

```python
'''problem5: Remove Invalid Entries.
A set contains valid course codes.
Remove a disconnected course code from the set.
Concept used: remove()/ discard()'''
course_codes={"EG101", "EG102", "EG103"}
course_codes.discard("EG102")
print(course_codes)
```
